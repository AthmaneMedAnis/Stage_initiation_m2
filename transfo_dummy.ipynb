```python
import pandas as pd
import numpy as np
import re
import ast
```

```python
df = pd.read_parquet("./data/train-00000-of-00001.parquet")
df.head()
```

```python
df_documents = df['documents'].apply(pd.Series)
```

```python
type(df_documents.loc[0, 'header'][0])
```

```python
def extraire_entier(x):
    return int(np.sum(x))

df_documents['word_count'] = df_documents['word_count'].apply(extraire_entier)
```

```python
def ndarray_tolist(x): 
    return x.tolist()
df_documents['header'] = df_documents['header'].apply(ndarray_tolist)
df_documents['text'] = df_documents['text'].apply(ndarray_tolist)
df_documents['type'] = df_documents['type'].apply(ndarray_tolist)
```
